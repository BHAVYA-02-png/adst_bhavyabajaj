```python
import pandas as pd

# Load the original dataset
df = pd.read_csv("Grocery_Inventory_and_Sales_Dataset.csv")

# Fix the category column name if required
if "Catagory" in df.columns:
    df.rename(columns={"Catagory": "Category"}, inplace=True)

# Fill the missing category for Cabbage
df.loc[df["Product_Name"] == "Cabbage", "Category"] = "Fruits & Vegetables"

# Convert date columns
df["Date_Received"] = pd.to_datetime(df["Date_Received"])
df["Last_Order_Date"] = pd.to_datetime(df["Last_Order_Date"])
df["Expiration_Date"] = pd.to_datetime(df["Expiration_Date"])

# Convert Unit_Price from text to numeric
df["Unit_Price"] = (
    df["Unit_Price"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .astype(float)
)

print("Dataset loaded and cleaned successfully!")
```

```python
df.to_csv("cleaned_supermarket_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
print("Earliest Date Received:", df["Date_Received"].min())
print("Latest Date Received:", df["Date_Received"].max())

print("Earliest Expiration Date:", df["Expiration_Date"].min())
print("Latest Expiration Date:", df["Expiration_Date"].max())
```

```python
df["Shelf_Life_Days"] = (
    df["Expiration_Date"] - df["Date_Received"]
).dt.days

print("Shelf life calculated successfully!")
```

```python
df["Overstock_Ratio"] = (
    df["Stock_Quantity"] / df["Sales_Volume"]
)

print("Overstock Ratio created successfully!")
```

```python
def classify_overstock(ratio):
    if ratio < 0.75:
        return "Low Risk"
    elif ratio <= 1.50:
        return "Medium Risk"
    else:
        return "High Risk"


df["Overstock_Risk"] = df["Overstock_Ratio"].apply(classify_overstock)

print("Overstock risk categories created successfully!")
```
